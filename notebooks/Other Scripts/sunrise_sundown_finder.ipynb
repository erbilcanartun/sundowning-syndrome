```python
import solarpy as sp
from datetime import datetime, timedelta
```

```python
### sunrise & sundown finder

def sunrise_sundown(date, latitude):
    
    counter = 0
    switch = "sunrise"
    for m in range(1440): # 1440 min = 24 h
    
        theta_s = sp.theta_z(date+timedelta(minutes=m),latitude) # zenith angle of sun
    
        if switch == "sundown" and counter < 10:
            counter += 1
            continue
        else: pass
    
        if switch == "sunrise":
            if round(theta_s,2) == 1.77:
                sunrise = m
                switch = "sundown"
            else: pass
    
        else:
            if round(theta_s,2) == 1.77:
                sundown = m
                break
            else: pass
    
    return timedelta(minutes=sunrise), timedelta(minutes=sundown)
```

```python
# run the function

date = datetime(2020,11,30)
latitude = 41

hour = sunrise_sundown(date,latitude)

print("sunrise: ", hour[0])
print("sundown: ", hour[1])
```
